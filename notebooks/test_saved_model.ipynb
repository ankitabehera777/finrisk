```python
import joblib
import pandas as pd

# Load saved files
model = joblib.load(r"C:\Users\HP\Documents\finrisk\models\finrisk_xgboost_model.pkl")
preprocessor = joblib.load(r"C:\Users\HP\Documents\finrisk\models\finrisk_preprocessor.pkl")
feature_names = joblib.load(r"C:\Users\HP\Documents\finrisk\models\finrisk_feature_names.pkl")

# Load the ML dataset
df = pd.read_csv(r"C:\Users\HP\Documents\finrisk\data\raw\finrisk_ml_dataset.csv")

# Select one loan
sample_loan = df[feature_names].iloc[[0]]

# Apply preprocessing
sample_loan_processed = preprocessor.transform(sample_loan)

# Predict
risk_probability = model.predict_proba(sample_loan_processed)[0][1]
predicted_class = int(risk_probability >= 0.5)

print("Saved model loaded successfully.")
print(f"Risk probability: {risk_probability:.4f}")
print(f"Risk probability percentage: {risk_probability * 100:.2f}%")
print(f"Predicted class: {predicted_class}")

if predicted_class == 1:
    print("Prediction: Higher-risk loan")
else:
    print("Prediction: Lower-risk loan")
```
